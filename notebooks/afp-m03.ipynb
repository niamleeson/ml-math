**Watch one logistic model learn by moving opposite the gradient.**

This notebook computes log loss and gradient-descent updates directly. _Save a copy to your Drive (File -> Save a copy in Drive) to keep your edits._

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(3)
```

## The key formula

For binary logistic regression, $p=\sigma(wx+b)$ and

$$\ell(p,y)=-[y\log p+(1-y)\log(1-p)]$$

The one-example weight gradient is $(p-y)x$.

```python
x = np.array([2.0])
y = np.array([1.0])
w = 0.3
b = -1.0
eta = 0.1

z = w * x + b
p = 1.0 / (1.0 + np.exp(-z))
loss = -(y * np.log(p) + (1.0 - y) * np.log(1.0 - p))

print("p", round(float(p[0]), 3))
print("loss", round(float(loss[0]), 3))
```

## Step 1 - Compute one gradient by hand in code

The positive label is underpredicted, so the gradient should increase the weight.

```python
grad_w = float((p - y) * x)
grad_b = float(p - y)

print("grad_w", round(grad_w, 3))
print("grad_b", round(grad_b, 3))

assert grad_w < 0
```

## Step 2 - Take one SGD step

The update $w \leftarrow w - \eta g$ moves opposite the gradient.

```python
w_new = w - eta * grad_w
b_new = b - eta * grad_b

p_old = float(p[0])
p_new = float(1.0 / (1.0 + np.exp(-(w_new * x[0] + b_new))))

print("old p", round(p_old, 3))
print("new p", round(p_new, 3))

assert p_new > p_old
```

## Step 3 - Train on a small synthetic dataset

Now repeat the same update many times and store the loss curve.

```python
n = 600
X = rng.normal(size=n)
true_p = 1.0 / (1.0 + np.exp(-(1.4 * X - 0.8)))
Y = (rng.random(n) < true_p).astype(float)

w_fit = 0.0
b_fit = 0.0
eta_fit = 0.3
losses = []

for step in range(80):
    pred = 1.0 / (1.0 + np.exp(-(w_fit * X + b_fit)))
    loss_value = -np.mean(Y * np.log(pred + 1e-12) + (1.0 - Y) * np.log(1.0 - pred + 1e-12))
    grad_w_fit = np.mean((pred - Y) * X)
    grad_b_fit = np.mean(pred - Y)
    w_fit = w_fit - eta_fit * grad_w_fit
    b_fit = b_fit - eta_fit * grad_b_fit
    losses.append(loss_value)

print("first loss", round(losses[0], 3))
print("last loss", round(losses[-1], 3))

assert losses[-1] < losses[0]
```

## Visualize optimization

A healthy loss curve trends downward even if individual stochastic updates may wiggle.

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(losses)
ax.set_xlabel("gradient step")
ax.set_ylabel("log loss")
ax.set_title("loss decreases under gradient descent")
plt.show()
```
