# Bernoulli distribution

Run every cell from top to bottom. Each cell prints intermediate values and draws a small visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(7)
plt.style.use('default')
```

## 1. Build a Bernoulli PMF

```python
p = 0.20
pmf = pd.DataFrame({'x': [0, 1], 'P(X=x)': [1 - p, p]})
print('p = probability of success =', p)
display(pmf)
```

## 2. Visualize the two possible outcomes

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(pmf['x'].astype(str), pmf['P(X=x)'], color=['#bdd7e7', '#3182bd'])
ax.set_title('Bernoulli PMF with p = 0.20')
ax.set_xlabel('x')
ax.set_ylabel('probability')
ax.set_ylim(0, 1)
plt.show()
```

## 3. Simulate many Bernoulli trials

```python
samples = np.random.binomial(n=1, p=p, size=200)
print('First 30 simulated labels:', samples[:30])
print('Sample mean:', samples.mean())
print('Theoretical mean:', p)
print('Theoretical variance:', p * (1 - p))
```

## 4. Watch the sample mean converge toward p

```python
running_mean = np.cumsum(samples) / np.arange(1, len(samples) + 1)
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(running_mean, label='running mean')
ax.axhline(p, color='black', linestyle='--', label='true p')
ax.set_title('Average of 0/1 samples estimates p')
ax.set_xlabel('number of trials')
ax.set_ylabel('running mean')
ax.legend()
plt.show()
```
