# Dropout

Run every cell from top to bottom. Each cell prints intermediate values and draws a small visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(7)
plt.style.use('default')
```

## 1. Apply one dropout mask by hand

```python
a = np.array([2.0, 4.0, 6.0, 8.0])
mask = np.array([1.0, 0.0, 1.0, 0.0])
p_keep = 0.5
dropped = mask * a
inverted = dropped / p_keep
print('activation a:', a)
print('mask:', mask)
print('mask * a:', dropped)
print('inverted dropout output:', inverted)
```

## 2. Visualize before and after dropout

```python
idx = np.arange(len(a))
fig, ax = plt.subplots(figsize=(6, 3))
ax.bar(idx - 0.2, a, width=0.4, label='original a')
ax.bar(idx + 0.2, inverted, width=0.4, label='after inverted dropout')
ax.set_xticks(idx)
ax.set_title('Dropout removes some activations and scales the kept ones')
ax.legend()
plt.show()
```

## 3. Simulate many masks to check the expected value

```python
runs = 5000
masks = np.random.binomial(n=1, p=p_keep, size=(runs, len(a)))
outputs = (masks * a) / p_keep
mean_output = outputs.mean(axis=0)
check = pd.DataFrame({'unit': np.arange(len(a)), 'original_activation': a, 'mean_after_many_masks': mean_output})
display(check)
print('Sanity check: inverted dropout keeps the average close to the original activation.')
```

## 4. Visualize the noisy training signal

```python
unit = 2
fig, ax = plt.subplots(figsize=(6, 3))
ax.hist(outputs[:, unit], bins=20, color='#9ecae1', edgecolor='black')
ax.axvline(a[unit], color='red', linestyle='--', label='original activation')
ax.set_title('One unit over many dropout masks')
ax.set_xlabel('output value during training')
ax.legend()
plt.show()
```
