# A* search

Run every cell from top to bottom. Each cell prints intermediate values and draws a small visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(7)
plt.style.use('default')
```

## 1. Define a tiny weighted graph and heuristic

```python
edges = {
    'S': {'A': 1, 'B': 4},
    'A': {'G': 5},
    'B': {'G': 1},
    'G': {}
}
h = {'S': 4, 'A': 4, 'B': 1, 'G': 0}
print('Edges:', edges)
print('Heuristic h:', h)
```

## 2. Run A* and record the frontier

```python
frontier = [{'node': 'S', 'path': ['S'], 'g': 0, 'h': h['S'], 'f': h['S']}]
trace = []
visited = set()

while frontier:
    frontier = sorted(frontier, key=lambda item: item['f'])
    current = frontier.pop(0)
    trace.append({'expanded': current['node'], 'path': ' -> '.join(current['path']), 'g': current['g'], 'h': current['h'], 'f': current['f'], 'frontier_after_pop': [x['node'] for x in frontier]})
    if current['node'] == 'G':
        best = current
        break
    if current['node'] in visited:
        continue
    visited.add(current['node'])
    for nxt, cost in edges[current['node']].items():
        g_new = current['g'] + cost
        frontier.append({'node': nxt, 'path': current['path'] + [nxt], 'g': g_new, 'h': h[nxt], 'f': g_new + h[nxt]})

trace_df = pd.DataFrame(trace)
display(trace_df)
print('Best path:', ' -> '.join(best['path']))
print('Best cost:', best['g'])
```

## 3. Draw the graph and highlight the chosen path

```python
pos = {'S': (0, 0), 'A': (1, 1), 'B': (1, -1), 'G': (2, 0)}
path_edges = set(zip(best['path'][:-1], best['path'][1:]))
fig, ax = plt.subplots(figsize=(6, 4))
for u, nbrs in edges.items():
    for v, cost in nbrs.items():
        x1, y1 = pos[u]
        x2, y2 = pos[v]
        color = '#de2d26' if (u, v) in path_edges else '#999999'
        width = 3 if (u, v) in path_edges else 1
        ax.plot([x1, x2], [y1, y2], color=color, linewidth=width)
        ax.text((x1 + x2) / 2, (y1 + y2) / 2, str(cost), fontsize=11)
for node, (x, y0) in pos.items():
    ax.scatter([x], [y0], s=700, color='#deebf7', edgecolor='black', zorder=3)
    ax.text(x, y0, node + '\n' + 'h=' + str(h[node]), ha='center', va='center', zorder=4)
ax.set_title('A* selected path: ' + ' -> '.join(best['path']))
ax.axis('off')
plt.show()
```

## 4. Inspect the f = g + h calculation for the first expansion

```python
first_options = pd.DataFrame([
    {'node': 'A', 'g': 1, 'h': h['A'], 'f': 1 + h['A']},
    {'node': 'B', 'g': 4, 'h': h['B'], 'f': 4 + h['B']}
])
display(first_options)
print('A* expands the smallest f value. Ties can be broken by implementation choice.')
```
