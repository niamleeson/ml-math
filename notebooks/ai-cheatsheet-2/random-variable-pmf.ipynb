# Random variable and PMF

Run every cell from top to bottom. Each cell prints intermediate values and draws a small visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(7)
plt.style.use('default')
```

## 1. Build the sample space
We roll one fair die, then define a random variable X that equals 1 for even rolls and 0 for odd rolls.

```python
omega = np.arange(1, 7)
df = pd.DataFrame({'die_roll': omega})
df['X_even_indicator'] = (df['die_roll'] % 2 == 0).astype(int)
df['probability'] = 1 / len(df)
print('Sample space with the random variable attached:')
display(df)
```

## 2. Compute the PMF from the table

```python
pmf = df.groupby('X_even_indicator', as_index=False)['probability'].sum()
pmf = pmf.rename(columns={'X_even_indicator': 'x', 'probability': 'p_X(x)'})
print('PMF table:')
display(pmf)
print('Sanity check: probabilities sum to', pmf['p_X(x)'].sum())
```

## 3. Visualize the PMF

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(pmf['x'].astype(str), pmf['p_X(x)'], color=['#9ecae1', '#3182bd'])
ax.set_title('PMF of X: odd -> 0, even -> 1')
ax.set_xlabel('value of X')
ax.set_ylabel('probability')
ax.set_ylim(0, 1)
plt.show()
```

## 4. Simulate rolls and compare empirical frequencies

```python
rolls = np.random.choice(omega, size=1000, replace=True)
X = (rolls % 2 == 0).astype(int)
sim = pd.Series(X).value_counts(normalize=True).sort_index().rename('simulated_probability').reset_index()
sim = sim.rename(columns={'index': 'x'})
print('First 20 rolls:', rolls[:20])
print('First 20 X values:', X[:20])
display(sim)
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(sim['x'].astype(str), sim['simulated_probability'], color=['#a1d99b', '#31a354'])
ax.axhline(0.5, color='black', linestyle='--', linewidth=1, label='true probability 0.5')
ax.set_title('Simulation approaches the PMF')
ax.set_ylim(0, 1)
ax.legend()
plt.show()
```
