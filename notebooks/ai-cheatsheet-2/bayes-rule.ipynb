# Bayes' rule

Run every cell from top to bottom. Each cell prints intermediate values and draws a small visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(7)
plt.style.use('default')
```

## 1. Define the disease-test scenario

```python
prior_disease = 0.01
sensitivity = 0.90
false_positive_rate = 0.05
prior_no_disease = 1 - prior_disease

print('P(disease) =', prior_disease)
print('P(positive | disease) =', sensitivity)
print('P(positive | no disease) =', false_positive_rate)
```

## 2. Compute total evidence probability

```python
true_positive_share = sensitivity * prior_disease
false_positive_share = false_positive_rate * prior_no_disease
evidence_positive = true_positive_share + false_positive_share
posterior = true_positive_share / evidence_positive

rows = pd.DataFrame({
    'source_of_positive_test': ['true positive: disease and positive', 'false positive: no disease but positive'],
    'probability_mass': [true_positive_share, false_positive_share]
})
display(rows)
print('P(positive) =', evidence_positive)
print('P(disease | positive) =', posterior)
```

## 3. Visualize why the posterior is not 90 percent

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.bar(rows['source_of_positive_test'], rows['probability_mass'], color=['#de2d26', '#9ecae1'])
ax.set_title('Most positive tests can still come from false positives when the disease is rare')
ax.set_ylabel('probability mass')
ax.tick_params(axis='x', rotation=15)
plt.show()
```

## 4. Explore different prevalences

```python
def bayes_update(prevalence, sensitivity=0.90, false_positive_rate=0.05):
    tp = sensitivity * prevalence
    fp = false_positive_rate * (1 - prevalence)
    return tp / (tp + fp)

prevalences = np.linspace(0.001, 0.5, 100)
posteriors = [bayes_update(p) for p in prevalences]
print('Posterior at 1% prevalence:', bayes_update(0.01))
print('Posterior at 20% prevalence:', bayes_update(0.20))
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(prevalences, posteriors)
ax.set_xlabel('prior prevalence P(disease)')
ax.set_ylabel('posterior P(disease | positive)')
ax.set_title('Bayes update as the prior changes')
ax.grid(True, alpha=0.3)
plt.show()
```

Try changing `false_positive_rate` to see why highly specific tests matter for rare events.
