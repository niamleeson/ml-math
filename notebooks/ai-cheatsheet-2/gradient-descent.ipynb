# Gradient descent

Run every cell from top to bottom. Each cell prints intermediate values and draws a small visualization.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(7)
plt.style.use('default')
```

## 1. Define a loss and its gradient

```python
def loss(theta):
    return (theta - 3) ** 2

def grad(theta):
    return 2 * (theta - 3)

theta = 0.0
alpha = 0.1
print('Starting theta:', theta)
print('Starting loss:', loss(theta))
print('Starting gradient:', grad(theta))
```

## 2. Run gradient descent and print every step

```python
history = []
for step in range(20):
    gradient = grad(theta)
    current_loss = loss(theta)
    history.append({'step': step, 'theta': theta, 'gradient': gradient, 'loss': current_loss})
    theta = theta - alpha * gradient

history_df = pd.DataFrame(history)
display(history_df.head(10))
print('Final theta after 20 updates:', theta)
print('Final loss:', loss(theta))
```

## 3. Visualize the parameter path on the loss curve

```python
grid = np.linspace(-1, 6, 200)
fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(grid, loss(grid), label='J(theta) = (theta - 3)^2')
ax.scatter(history_df['theta'], history_df['loss'], color='#de2d26', label='gradient descent steps')
for i, row in history_df.iloc[:6].iterrows():
    ax.annotate(str(int(row['step'])), (row['theta'], row['loss']), textcoords='offset points', xytext=(4, 4), fontsize=8)
ax.set_xlabel('theta')
ax.set_ylabel('loss')
ax.set_title('Gradient descent walks downhill')
ax.legend()
plt.show()
```

## 4. Compare learning rates

```python
def run(alpha, steps=20):
    theta = 0.0
    losses = []
    for step in range(steps):
        losses.append(loss(theta))
        theta = theta - alpha * grad(theta)
    return losses

fig, ax = plt.subplots(figsize=(6, 3))
for alpha in [0.02, 0.1, 0.8, 1.05]:
    ax.plot(run(alpha), label=f'alpha={alpha}')
ax.set_title('Learning rate changes convergence')
ax.set_xlabel('step')
ax.set_ylabel('loss')
ax.set_yscale('log')
ax.legend()
plt.show()
```
