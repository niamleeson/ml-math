# Vector Math Drills — self-checking notebook

Every formula from `vector-math-formulas.md` with 5 basic, 5 easy, and 5 advanced questions. Each formula has a markdown cell with the questions and step-by-step solutions, followed by a code cell that recomputes and **asserts** every answer.

Run all cells: each formula prints `Formula X.Y: N checks passed`.

```python
import numpy as np, math

def edit_distance(s, t):
    m, n = len(s), len(t)
    dp = [[0]*(n+1) for _ in range(m+1)]
    for i in range(m+1): dp[i][0] = i
    for j in range(n+1): dp[0][j] = j
    for i in range(1, m+1):
        for j in range(1, n+1):
            c = 0 if s[i-1]==t[j-1] else 1
            dp[i][j] = min(dp[i-1][j]+1, dp[i][j-1]+1, dp[i-1][j-1]+c)
    return dp[m][n]

print("helpers ready")
```

## Formula 1.1 — Vector addition

**Basic 1.** Compute [2, 2] + [1, 3].

Solution:
- Step 1: Add component 1: 2 + (1) = 3
- Step 2: Add component 2: 2 + (3) = 5

**Answer:** [3, 5]

**Basic 2.** Compute [3, 2] + [4, 1].

Solution:
- Step 1: Add component 1: 3 + (4) = 7
- Step 2: Add component 2: 2 + (1) = 3

**Answer:** [7, 3]

**Basic 3.** Compute [2, 1] + [3, 2].

Solution:

## Formula 1.5 — Linear combination

**Basic 1.** Compute 3 * [3, 2] + 3 * [2, 4].

Solution:
- Step 1: Scale the first vector by 3: 3 * [3, 2] = [9, 6]
- Step 2: Scale the second vector by 3: 3 * [2, 4] = [6, 12]
- Step 3: Add component 1: 9 + (6) = 15
- Step 4: Add component 2: 6 + (12) = 18

**Answer:** [15, 18]

Solution:
- Step 1: Scale the first vector by 3: 3 * [2, 1] = [6, 3]

## Formula 2.4 — Scaling rule for norms

## Formula 3.5 — Dot product is commutative

**Basic 1.** Verify dot(a,b) = dot(b,a) for a = [1, 2], b = [2, 3].

Solution:
- Step 1: dot(a,b): Multiply component 1: 1 * 2 = 2
- Step 2: dot(a,b): Multiply component 2: 2 * 3 = 6
- Step 3: dot(a,b): Add the products: 2 + 6 = 8
- Step 4: dot(b,a): Multiply component 1: 2 * 1 = 2
- Step 5: dot(b,a): Multiply component 2: 3 * 2 = 6
- Step 6: dot(b,a): Add the products: 2 + 6 = 8
- Step 7: Both equal 8, so order does not matter.

**Answer:** both = 8

**Basic 2.** Verify dot(a,b) = dot(b,a) for a = [3, 4], b = [1, 3].

Solution:
- Step 1: dot(a,b): Multiply component 1: 3 * 1 = 3
- Step 2: dot(a,b): Multiply component 2: 4 * 3 = 12
- Step 3: dot(a,b): Add the products: 3 + 12 = 15
- Step 4: dot(b,a): Multiply component 1: 1 * 3 = 3
- Step 5: dot(b,a): Multiply component 2: 3 * 4 = 12
- Step 6: dot(b,a): Add the products: 3 + 12 = 15
- Step 7: Both equal 15, so order does not matter.

**Answer:** both = 15

**Basic 3.** Verify dot(a,b) = dot(b,a) for a = [3, 5], b = [4, 1].

Solution:
- Step 1: dot(a,b): Multiply component 1: 3 * 4 = 12
- Step 2: dot(a,b): Multiply component 2: 5 * 1 = 5

## Formula 4.1 — Cosine similarity

**Basic 1.** Compute the cosine similarity of [2, 2] and [5, 2].

Solution:
- Step 1: dot: Multiply component 1: 2 * 5 = 10
- Step 2: dot: Multiply component 2: 2 * 2 = 4
- Step 3: dot: Add the products: 10 + 4 = 14
- Step 4: norm(a): Square component 1: (2)^2 = 4
- Step 5: norm(a): Square component 2: (2)^2 = 4
- Step 6: norm(a): Add the squares: 4 + 4 = 8
- Step 7: norm(a): Take the square root: sqrt(8) = 2.8284
- Step 8: norm(b): Square component 1: (5)^2 = 25
- Step 9: norm(b): Square component 2: (2)^2 = 4
- Step 10: norm(b): Add the squares: 25 + 4 = 29
- Step 11: norm(b): Take the square root: sqrt(29) = 5.3852
- Step 12: Multiply the norms: 2.8284 * 5.3852 = 15.2315.
- Step 13: Divide dot by that: 14 / 15.2315 = 0.9191.

**Answer:** 0.9191

## Formula 4.2 — Cosine distance

**Basic 1.** Compute the cosine distance (1 - cosine) of [5, 1] and [2, 3].

Solution:
- Step 1: dot: Multiply component 1: 5 * 2 = 10
- Step 2: dot: Multiply component 2: 1 * 3 = 3
- Step 3: dot: Add the products: 10 + 3 = 13
- Step 4: norm(a): Square component 1: (5)^2 = 25
- Step 5: norm(a): Square component 2: (1)^2 = 1
- Step 6: norm(a): Add the squares: 25 + 1 = 26
- Step 7: norm(a): Take the square root: sqrt(26) = 5.099
- Step 8: norm(b): Square component 1: (2)^2 = 4
- Step 9: norm(b): Square component 2: (3)^2 = 9
- Step 10: norm(b): Add the squares: 4 + 9 = 13
- Step 11: norm(b): Take the square root: sqrt(13) = 3.6056
- Step 12: Multiply the norms: 5.099 * 3.6056 = 18.3848.
- Step 13: Cosine similarity: 13 / 18.3848 = 0.7071.
- Step 14: Cosine distance: 1 - 0.7071 = 0.2929.

**Answer:** 0.2929

## Formula 4.3 — Angle from cosine

**Basic 1.** Find the angle (in degrees) between [5, 2] and [2, 3].

Solution:
- Step 1: dot: Multiply component 1: 5 * 2 = 10
- Step 2: dot: Multiply component 2: 2 * 3 = 6
- Step 3: dot: Add the products: 10 + 6 = 16
- Step 4: norm(a): Square component 1: (5)^2 = 25
- Step 5: norm(a): Square component 2: (2)^2 = 4
- Step 6: norm(a): Add the squares: 25 + 4 = 29
- Step 7: norm(a): Take the square root: sqrt(29) = 5.3852
- Step 8: norm(b): Square component 1: (2)^2 = 4
- Step 9: norm(b): Square component 2: (3)^2 = 9
- Step 10: norm(b): Add the squares: 4 + 9 = 13
- Step 11: norm(b): Take the square root: sqrt(13) = 3.6056
- Step 12: Cosine similarity: 16 / (5.3852 * 3.6056) = 0.824.
- Step 13: Take the inverse cosine: arccos(0.824) = 34.5085 degrees.

**Answer:** 34.5085

## Formula 4.4 — Cauchy-Schwarz inequality

**Basic 1.** Verify the Cauchy-Schwarz inequality |dot| <= norm*norm for a = [5, 3], b = [5, 5].

Solution:
- Step 1: dot: Multiply component 1: 5 * 5 = 25
- Step 2: dot: Multiply component 2: 3 * 5 = 15
- Step 3: dot: Add the products: 25 + 15 = 40
- Step 4: Absolute value of the dot product: |40| = 40.
- Step 5: norm(a): Square component 1: (5)^2 = 25
- Step 6: norm(a): Square component 2: (3)^2 = 9
- Step 7: norm(a): Add the squares: 25 + 9 = 34
- Step 8: norm(a): Take the square root: sqrt(34) = 5.831
- Step 9: norm(b): Square component 1: (5)^2 = 25
- Step 10: norm(b): Square component 2: (5)^2 = 25
- Step 11: norm(b): Add the squares: 25 + 25 = 50
- Step 12: norm(b): Take the square root: sqrt(50) = 7.0711
- Step 13: Multiply the norms: 5.831 * 7.0711 = 41.2311.
- Step 14: Check: 40 <= 41.2311 is True.

**Answer:** 40 <= 41.2311 (true)

## Formula 5.2 — Dot of two unit vectors = cosine

**Basic 1.** Normalize a = [8, 6] and b = [8, 6], then take their dot product (this equals the cosine).

Solution:
- Step 1: norm(a): Square component 1: (8)^2 = 64
- Step 2: norm(a): Square component 2: (6)^2 = 36
- Step 3: norm(a): Add the squares: 64 + 36 = 100
- Step 4: norm(a): Take the square root: sqrt(100) = 10
- Step 5: Unit vector a_hat = [0.8, 0.6].
- Step 6: norm(b): Square component 1: (8)^2 = 64
- Step 7: norm(b): Square component 2: (6)^2 = 36
- Step 8: norm(b): Add the squares: 64 + 36 = 100
- Step 9: norm(b): Take the square root: sqrt(100) = 10
- Step 10: Unit vector b_hat = [0.8, 0.6].
- Step 11: dot of unit vectors: Multiply component 1: 0.8 * 0.8 = 0.64
- Step 12: dot of unit vectors: Multiply component 2: 0.6 * 0.6 = 0.36
- Step 13: dot of unit vectors: Add the products: 0.64 + 0.36 = 1

**Answer:** 1

## Formula 5.3 — Mean-centering

**Basic 1.** Mean-center the vector [2, 4] (subtract its average from each entry).

Solution:
- Step 1: Add the entries: 2 + 4 = 6.
- Step 2: Divide by the count 2: 6 / 2 = 3.
- Step 3: Subtract the mean from component 1: 2 - 3 = -1
- Step 4: Subtract the mean from component 2: 4 - 3 = 1

**Answer:** [-1, 1]

## Formula 5.4 — Standardization (z-score)

**Basic 1.** Standardize [4, 3] (subtract mean, divide by standard deviation).

Solution:
- Step 1: Mean: (4 + 3) / 2 = 3.5.
- Step 2: Squared deviations: (4-3.5)^2=0.25, (3-3.5)^2=0.25.
- Step 3: Variance: average of those = 0.25.
- Step 4: Standard deviation: sqrt(0.25) = 0.5.
- Step 5: z of component 1: (4 - 3.5) / 0.5 = 1
- Step 6: z of component 2: (3 - 3.5) / 0.5 = -1

**Answer:** [1, -1]

**Basic 2.** Standardize [1, 2] (subtract mean, divide by standard deviation).

Solution:
- Step 1: Mean: (1 + 2) / 2 = 1.5.
- Step 2: Squared deviations: (1-1.5)^2=0.25, (2-1.5)^2=0.25.
- Step 3: Variance: average of those = 0.25.
- Step 4: Standard deviation: sqrt(0.25) = 0.5.
- Step 5: z of component 1: (1 - 1.5) / 0.5 = -1
- Step 6: z of component 2: (2 - 1.5) / 0.5 = 1

**Answer:** [-1, 1]

**Basic 3.** Standardize [4, 2] (subtract mean, divide by standard deviation).

Solution:
- Step 1: Mean: (4 + 2) / 2 = 3.
- Step 2: Squared deviations: (4-3)^2=1, (2-3)^2=1.
- Step 3: Variance: average of those = 1.
- Step 4: Standard deviation: sqrt(1) = 1.
- Step 5: z of component 1: (4 - 3) / 1 = 1
- Step 6: z of component 2: (2 - 3) / 1 = -1

**Answer:** [1, -1]

**Basic 4.** Standardize [4, 2] (subtract mean, divide by standard deviation).

Solution:

## Formula 6.1 — Euclidean distance

**Basic 1.** Compute the Euclidean distance between [5, 1] and [2, 3].

Solution:
- Step 1: Subtract component 1: 5 - (2) = 3
- Step 2: Subtract component 2: 1 - (3) = -2
- Step 3: Now take the norm of the difference [3, -2]:
- Step 4: Square component 1: (3)^2 = 9
- Step 5: Square component 2: (-2)^2 = 4
- Step 6: Add the squares: 9 + 4 = 13
- Step 7: Take the square root: sqrt(13) = 3.6056

**Answer:** 3.6056

## Formula 6.3 — Manhattan (L1) distance

## Formula 6.4 — Chebyshev (L-infinity) distance

**Basic 1.** Compute the Chebyshev (max) distance between [1, 5] and [5, 3].

Solution:
- Step 1: Subtract component 1: 1 - (5) = -4
- Step 2: Subtract component 2: 5 - (3) = 2
- Step 3: Absolute value of difference 1: |-4| = 4
- Step 4: Absolute value of difference 2: |2| = 2
- Step 5: Take the maximum: max(4, 2) = 4

**Answer:** 4

**Basic 2.** Compute the Chebyshev (max) distance between [5, 2] and [5, 2].

Solution:
- Step 1: Subtract component 1: 5 - (5) = 0
- Step 2: Subtract component 2: 2 - (2) = 0
- Step 3: Absolute value of difference 1: |0| = 0
- Step 4: Absolute value of difference 2: |0| = 0

## Formula 6.5 — Minkowski distance (p=3)

**Basic 1.** Compute the Minkowski distance (p=3) between [5, 1] and [4, 2].

Solution:
- Step 1: Subtract component 1: 5 - (4) = 1
- Step 2: Subtract component 2: 1 - (2) = -1
- Step 3: Raise |difference 1| to power 3: |1|^3 = 1
- Step 4: Raise |difference 2| to power 3: |-1|^3 = 1
- Step 5: Add them: 1 + 1 = 2
- Step 6: Take the 3rd root: (2)^(1/3) = 1.2599

**Answer:** 1.2599

## Formula 6.6 — Distance is symmetric

**Basic 1.** Verify distance(a,b) = distance(b,a) for a = [1, 2], b = [4, 4].

- Step 2: Distance(b,a) = norm([3, 2]) = 3.6056.
- Step 3: They are equal, so distance is symmetric.

**Answer:** both = 3.6056

## Formula 7.1 — Law of cosines (distance <-> dot)

**Basic 1.** Use the law of cosines to find the squared distance between a = [4, 5] and b = [5, 1].

Solution:
- Step 1: norm(a)^2 = dot(a,a) = 41.
- Step 2: norm(b)^2 = dot(b,b) = 26.
- Step 3: dot(a,b) = 25.
- Step 4: Apply distance^2 = norm(a)^2 + norm(b)^2 - 2*dot(a,b):
- Step 5:   = 41 + 26 - 2*25 = 17.

**Answer:** 17

## Formula 7.2 — Unit-sphere distance from cosine

**Basic 1.** Two unit vectors have cosine similarity -1. Find the squared distance between them.

Solution:
- Step 1: On unit vectors, distance^2 = 2 - 2*cosine.
- Step 2: Insert cosine = -1: 2 - 2*-1 = 4.
- Step 3: (Distance itself = sqrt(4) = 2.)

**Answer:** 4

## Formula 7.3 — Polarization identity

**Basic 1.** Use the polarization identity to recover dot(a,b) from lengths only, for a = [4, 1], b = [2, 1].

Solution:
- Step 1: a + b = [6, 2].
- Step 2: norm(a+b)^2 = 40.
- Step 3: norm(a)^2 = 17.
- Step 4: norm(b)^2 = 5.
- Step 5: Apply (norm(a+b)^2 - norm(a)^2 - norm(b)^2)/2:
- Step 6:   = (40 - 17 - 5) / 2 = 9.

**Answer:** 9

## Formula 8.1 — Projection length

**Basic 1.** Compute the projection length of a = [2, 1] onto b = [2, 5] (dot(a,b) / norm(b)).

Solution:
- Step 1: dot(a,b): Multiply component 1: 2 * 2 = 4
- Step 2: dot(a,b): Multiply component 2: 1 * 5 = 5
- Step 3: dot(a,b): Add the products: 4 + 5 = 9

## Formula 8.2 — Projection vector

**Basic 1.** Compute the projection vector of a = [1, 4] onto b = [5, 2].

Solution:
- Step 1: dot(a,b): Multiply component 1: 1 * 5 = 5
- Step 2: dot(a,b): Multiply component 2: 4 * 2 = 8
- Step 3: dot(a,b): Add the products: 5 + 8 = 13
- Step 4: dot(b,b): Multiply component 1: 5 * 5 = 25
- Step 5: dot(b,b): Multiply component 2: 2 * 2 = 4
- Step 6: dot(b,b): Add the products: 25 + 4 = 29
- Step 7: Ratio dot(a,b)/dot(b,b) = 13 / 29 = 0.4483.
- Step 8: Scale b by 0.4483:
- Step 9:   Component 1: 0.4483 * 5 = 2.2414
- Step 10:   Component 2: 0.4483 * 2 = 0.8966

**Answer:** [2.2414, 0.8966]

## Formula 9.3 — Cross product (3D)

**Basic 1.** Compute the cross product [5, 1, 5] x [2, 3, 5].

Solution:
- Step 1: x-component: a2*b3 - a3*b2 = 1*5 - 5*3 = -10
- Step 2: y-component: a3*b1 - a1*b3 = 5*2 - 5*5 = -15
- Step 3: z-component: a1*b2 - a2*b1 = 5*3 - 1*2 = 13

**Answer:** [-10, -15, 13]

## Formula 10.2 — Correlation = cosine of centered vectors

**Basic 1.** Compute the Pearson correlation of a = [7, 4, 1] and b = [5, 6, 3] (cosine of the centered vectors).

Solution:
- Step 1: Mean of a = 4; centered a = [3, 0, -3].
- Step 2: Mean of b = 4.6667; centered b = [0.3333, 1.3333, -1.6667].
- Step 3: dot(centered a, centered b) = 6.
- Step 4: norm(centered a) = 4.2426, norm(centered b) = 2.1602.
- Step 5: Correlation = 6 / (4.2426 * 2.1602) = 0.6547.

**Answer:** 0.6547

**Basic 2.** Compute the Pearson correlation of a = [3, 2, 4] and b = [7, 4, 6] (cosine of the centered vectors).

Solution:
- Step 1: Mean of a = 3; centered a = [0, -1, 1].

## Formula 10.4 — Covariance

**Basic 1.** Compute the covariance of a = [5, 6, 1] and b = [7, 1, 5].

Solution:
- Step 1: Mean of a = 4, mean of b = 4.3333.
- Step 2: Product 1: (5 - 4) * (7 - 4.3333) = 2.6667
- Step 3: Product 2: (6 - 4) * (1 - 4.3333) = -6.6667
- Step 4: Product 3: (1 - 4) * (5 - 4.3333) = -2
- Step 5: Average the products: (2.6667 + -6.6667 + -2) / 3 = -2

**Answer:** -2

**Basic 2.** Compute the covariance of a = [3, 1, 4] and b = [2, 4, 2].

Solution:
- Step 1: Mean of a = 2.6667, mean of b = 2.6667.
- Step 2: Product 1: (3 - 2.6667) * (2 - 2.6667) = -0.2222
- Step 3: Product 2: (1 - 2.6667) * (4 - 2.6667) = -2.2222
- Step 4: Product 3: (4 - 2.6667) * (2 - 2.6667) = -0.8889
- Step 5: Average the products: (-0.2222 + -2.2222 + -0.8889) / 3 = -1.1111

**Answer:** -1.1111

**Basic 3.** Compute the covariance of a = [1, 5, 1] and b = [3, 7, 7].

Solution:
- Step 1: Mean of a = 2.3333, mean of b = 5.6667.
- Step 2: Product 1: (1 - 2.3333) * (3 - 5.6667) = 3.5556
- Step 3: Product 2: (5 - 2.3333) * (7 - 5.6667) = 3.5556
- Step 4: Product 3: (1 - 2.3333) * (7 - 5.6667) = -1.7778
- Step 5: Average the products: (3.5556 + 3.5556 + -1.7778) / 3 = 1.7778

**Answer:** 1.7778

**Basic 4.** Compute the covariance of a = [1, 6, 2] and b = [4, 2, 3].

Solution:
- Step 1: Mean of a = 3, mean of b = 3.

- Step 3: Product 2: (6 - 3) * (2 - 3) = -3
- Step 4: Product 3: (2 - 3) * (3 - 3) = 0
- Step 5: Average the products: (-2 + -3 + 0) / 3 = -1.6667

**Answer:** -1.6667

**Basic 5.** Compute the covariance of a = [5, 2, 7] and b = [6, 2, 3].

- Step 4: Product 3: (7 - 4.6667) * (3 - 3.6667) = -1.5556
- Step 5: Average the products: (0.7778 + 4.4444 + -1.5556) / 3 = 1.2222

**Answer:** 1.2222

## Formula 11.1 — Linear interpolation (lerp)

Solution:
- Step 1: Formula: (1 - t)*a + t*b with t = 0.75.
- Step 2: 1 - t = 0.25.
- Step 3: Component 1: 0.25*5 + 0.75*1 = 2
- Step 4: Component 2: 0.25*3 + 0.75*1 = 1.5

## Formula 11.3 — 2D rotation

Solution:
- Step 1: cos(90) = 0, sin(90) = 1.
- Step 2: x_new = x*cos - y*sin = 7*0 - 2*1 = -2
- Step 3: y_new = x*sin + y*cos = 7*1 + 2*0 = 7

**Answer:** [-2, 7]

## Formula 11.4 — Reflection across a unit direction

**Basic 1.** Reflect the vector [3, 2] across the unit direction [1, 0].

Solution:
- Step 1: dot(v, u) = 3.
- Step 2: Formula: 2*dot(v,u)*u - v.
- Step 3: 2*3*u = [6, 0].
- Step 4: Subtract v: [6, 0] - [3, 2] = [3, -2].

**Answer:** [3, -2]

## Formula 14.6 — Inverse of a 2x2

- Step 2: Swap a and d, negate b and c: [[1, -5], [-4, 2]].
- Step 3: Divide every entry by the determinant -18:
- Step 4:   [[1/-18, -5/-18], [-4/-18, 2/-18]] = [[-0.0556, 0.2778]; [0.2222, -0.1111]].

## Formula 14.11 — Reconstruct a matrix from SVD factors

**Basic 1.** Reconstruct M = U S V^T for U = [[0, 0]; [1, 2]], S = [[1, 0]; [0, 2]], V^T = [[2, 1]; [1, 1]].

Solution:
- Step 1: First multiply U by S: [[0, 0]; [1, 2]] times [[1, 0]; [0, 2]] = [[0, 0]; [1, 4]].
- Step 2: Then multiply that by V^T: [[0, 0]; [1, 4]] times [[2, 1]; [1, 1]] = [[0, 0]; [6, 5]].

## Formula 14.16 — Gram matrix

**Basic 1.** Compute the 2x2 Gram matrix of a = [6, 3] and b = [3, 5] (all pairwise dot products).

Solution:
- Step 1: Entry (1,1) = dot(a,a) = 45.
- Step 2: Entry (1,2) = entry (2,1) = dot(a,b) = 33.
- Step 3: Entry (2,2) = dot(b,b) = 34.
- Step 4: Gram matrix = [[45, 33]; [33, 34]].

**Answer:** [[45, 33]; [33, 34]]

**Basic 2.** Compute the 2x2 Gram matrix of a = [1, 4] and b = [6, 1] (all pairwise dot products).

Solution:
- Step 1: Entry (1,1) = dot(a,a) = 17.
- Step 2: Entry (1,2) = entry (2,1) = dot(a,b) = 10.
- Step 3: Entry (2,2) = dot(b,b) = 37.
- Step 4: Gram matrix = [[17, 10]; [10, 37]].

**Answer:** [[17, 10]; [10, 37]]

**Basic 3.** Compute the 2x2 Gram matrix of a = [5, 2] and b = [3, 5] (all pairwise dot products).

Solution:
- Step 1: Entry (1,1) = dot(a,a) = 29.
- Step 2: Entry (1,2) = entry (2,1) = dot(a,b) = 25.
- Step 3: Entry (2,2) = dot(b,b) = 34.
- Step 4: Gram matrix = [[29, 25]; [25, 34]].

**Answer:** [[29, 25]; [25, 34]]

**Basic 4.** Compute the 2x2 Gram matrix of a = [5, 1] and b = [3, 6] (all pairwise dot products).

Solution:
- Step 1: Entry (1,1) = dot(a,a) = 26.

## Formula 15.1 — Gradient of a simple function

**Basic 1.** Find the gradient of f(x,y) = 2x^2 + 2y^2 at the point (1, 4).

Solution:
- Step 1: Partial with respect to x: d/dx of 2x^2 = 4x; at x=1 that is 4.
- Step 2: Partial with respect to y: d/dy of 2y^2 = 4y; at y=4 that is 16.
- Step 3: Gradient = [4, 16].

## Formula 15.3 — Gradient of the squared norm

**Basic 1.** Find the gradient of f(v) = norm(v)^2 at v = [5, 3].

Solution:
- Step 1: The gradient of norm(v)^2 is 2*v.
- Step 2: Component 1: 2 * 5 = 10
- Step 3: Component 2: 2 * 3 = 6

**Answer:** [10, 6]

**Basic 2.** Find the gradient of f(v) = norm(v)^2 at v = [3, 4].

Solution:
- Step 1: The gradient of norm(v)^2 is 2*v.

## Formula 15.5 — Directional derivative

**Basic 1.** The gradient of f is [3, 7, 3]. Find the directional derivative along the unit direction [1, 0, 0].

Solution:
- Step 1: Directional derivative = dot(gradient, direction).
- Step 2: Multiply component 1: 3 * 1 = 3
- Step 3: Multiply component 2: 7 * 0 = 0
- Step 4: Multiply component 3: 3 * 0 = 0
- Step 5: Add the products: 3 + 0 + 0 = 3

**Answer:** 3

## Formula 15.6 — Jacobian

**Basic 1.** Find the Jacobian of f(x,y) = [x^2 + y, x*y] at (4, 3).

Solution:
- Step 1: f1 = x^2 + y. d f1/dx = 2x = 8; d f1/dy = 1.
- Step 2: f2 = x*y. d f2/dx = y = 3; d f2/dy = x = 4.
- Step 3: Jacobian = [[8, 1]; [3, 4]].

**Answer:** [[8, 1]; [3, 4]]

## Formula 15.7 — Chain rule

**Basic 1.** Use the chain rule to differentiate f(x) = (3x + 4)^2 at x = 3.

Solution:
- Step 1: Inner function: u = g(x) = 3x + 4; at x=3, u = 13.
- Step 2: Outer derivative: f(u) = u^2, so f'(u) = 2u = 26.
- Step 3: Inner derivative: g'(x) = 3.
- Step 4: Multiply: 26 * 3 = 78.

**Answer:** 78

## Formula 15.8 — Gradient descent step

**Basic 1.** Take one gradient-descent step from w = [1, 2, 1] with gradient [3, 2, 1] and learning rate 0.5.

Solution:
- Step 1: Update rule: w_new = w - learning_rate * gradient, learning_rate = 0.5.
- Step 2: Component 1: 1 - 0.5*3 = -0.5
- Step 3: Component 2: 2 - 0.5*2 = 1
- Step 4: Component 3: 1 - 0.5*1 = 0.5

**Answer:** [-0.5, 1, 0.5]

**Basic 2.** Take one gradient-descent step from w = [7, 6, 3] with gradient [6, 4, 7] and learning rate 0.1.

Solution:
- Step 1: Update rule: w_new = w - learning_rate * gradient, learning_rate = 0.1.
- Step 2: Component 1: 7 - 0.1*6 = 6.4
- Step 3: Component 2: 6 - 0.1*4 = 5.6
- Step 4: Component 3: 3 - 0.1*7 = 2.3

**Answer:** [6.4, 5.6, 2.3]

## Formula 16.1 — Mahalanobis distance (diagonal covariance)

**Basic 1.** Compute the Mahalanobis distance between a = [5, 1] and b = [4, 4] with diagonal variances [4, 9].

Solution:
- Step 1: Differences: 5-4=1, 1-4=-3.
- Step 2: Variances (diagonal covariance): [4, 9].
- Step 3: Term 1: (1)^2 / 4 = 0.25
- Step 4: Term 2: (-3)^2 / 9 = 1
- Step 5: Add the terms: 0.25 + 1 = 1.25
- Step 6: Take the square root: sqrt(1.25) = 1.118

**Answer:** 1.118

**Basic 2.** Compute the Mahalanobis distance between a = [4, 3] and b = [5, 1] with diagonal variances [4, 9].

Solution:
- Step 1: Differences: 4-5=-1, 3-1=2.
- Step 2: Variances (diagonal covariance): [4, 9].
- Step 3: Term 1: (-1)^2 / 4 = 0.25
- Step 4: Term 2: (2)^2 / 9 = 0.4444
- Step 5: Add the terms: 0.25 + 0.4444 = 0.6944
- Step 6: Take the square root: sqrt(0.6944) = 0.8333

**Answer:** 0.8333

**Basic 3.** Compute the Mahalanobis distance between a = [5, 1] and b = [5, 4] with diagonal variances [9, 9].

Solution:
- Step 1: Differences: 5-5=0, 1-4=-3.

- Step 3: Term 1: (0)^2 / 9 = 0
- Step 4: Term 2: (-3)^2 / 9 = 1
- Step 5: Add the terms: 0 + 1 = 1
- Step 6: Take the square root: sqrt(1) = 1

**Answer:** 1

## Formula 16.7 — RBF / Gaussian kernel

## Formula 17.1 — Softmax

**Basic 1.** Compute the softmax of [1, 1].

- Step 1: Exponentiate component 1: exp(1) = 2.7183

## Formula 17.4 — KL divergence

**Basic 1.** Compute the KL divergence KL(p || q) for p = [0.5, 0.5], q = [0.25, 0.75].

Solution:
- Step 1: Term 1: 0.5 * ln(0.5/0.25) = 0.5 * ln(2) = 0.3466
- Step 2: Term 2: 0.5 * ln(0.5/0.75) = 0.5 * ln(0.6667) = -0.2027
- Step 3: Add the terms: 0.3466 + -0.2027 = 0.1438

**Answer:** 0.1438

**Basic 2.** Compute the KL divergence KL(p || q) for p = [0.9, 0.1], q = [0.8, 0.2].

Solution:
- Step 1: Term 1: 0.9 * ln(0.9/0.8) = 0.9 * ln(1.125) = 0.106
- Step 2: Term 2: 0.1 * ln(0.1/0.2) = 0.1 * ln(0.5) = -0.0693
- Step 3: Add the terms: 0.106 + -0.0693 = 0.0367

**Answer:** 0.0367
